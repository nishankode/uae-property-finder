```python
import os
import pandas as pd
from pathlib import Path

def get_sheet_names(base_path):
    base_path = Path(base_path)
    
    if not base_path.is_dir():
        raise NotADirectoryError(f"{base_path} is not a valid directory.")

    res = {}

    for file_path in base_path.glob("*.xls*"):
        try:
            xls = pd.ExcelFile(file_path)
            res[file_path.name] = xls.sheet_names
        except Exception as e:
            print(f"Could not read {file_path.name}: {e}")

    return res
```

```python
base_path = '../data/owner_data'
available_sheets = get_sheet_names(base_path)
```

```python
all_s1_data = []
base_path = '../data/owner_data'
for file, sheet in zip(sheet_availability.loc[:, 'file_name'], sheet_availability.loc[:, 'second_sheet']):
	file_path = os.path.join(base_path, file)
	data = pd.read_excel(file_path, sheet_name=sheet)
	data['file_name'] = file
	all_s1_data.append(data)
```

```python
import pandas as pd
import numpy as np

def create_unified_data(df1, df2):
    # 1. Define the Semantic Map
    # Key = data1 column name, Value = data2 column name
    mapping = {
        'Regis': 'Date',
        'ProcedureValue': 'Transaction Amount',
        'Master Project Land': 'Master Location',
        'Master Project': 'Master Project',
        'Building No': 'Building 1',
        'BuildingNameEn': 'BuildingName 2',
        'UnitNumber': 'property_number',
        'PropertyTypeEn': 'Property Type',
        'Size': 'Actual Size',
        'NameEn': 'Owner Name',
        'Mobile': 'Mobile 2'
    }
    
    # 2. Prepare data1
    # Rename and drop columns that don't exist in the target schema
    df1_mapped = df1.rename(columns=mapping)
    
    # 3. Handle Data Types
    # Ensure Dates are compatible
    df1_mapped['Date'] = pd.to_datetime(df1_mapped['Date'], errors='coerce')
    df2['Date'] = pd.to_datetime(df2['Date'], errors='coerce')
    
    # Ensure Numeric values are floats (stripping potential formatting if needed)
    for col in ['Transaction Amount', 'Actual Size']:
        df1_mapped[col] = pd.to_numeric(df1_mapped[col], errors='coerce')
        df2[col] = pd.to_numeric(df2[col], errors='coerce')

    # 4. Align Schemas
    # Force df1 to have all columns present in df2 (even if they are empty like 'beds')
    # and remove extra columns from df1 (like 'DmSubNo' or 'IdNumber')
    df1_aligned = df1_mapped.reindex(columns=df2.columns)
    
    # 5. Concatenate
    data3 = pd.concat([df2, df1_aligned], axis=0, ignore_index=True)
    
    return data3

# Execution
data3 = create_unified_data(old, df)
```
